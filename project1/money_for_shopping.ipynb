```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Data Processing and Exploration

```python
data_path = "./data/project_statistics_data.csv"
df_data = pd.read_csv(data_path)
df_data.head()
```

```python
df_data.describe()
```

```python
df_data.info()
```

```python
df_data.dtypes
```

There is one row having no annual income data. We chose to remove this observation.

```python
df_data = df_data.dropna().reset_index(drop=True)
df_data.head()
```

```python
cols = ["Age", "Annual Income", "Normalized spending"]

fig, axes = plt.subplots(1, 3, figsize=(12, 4), constrained_layout=True)
axes = axes.flatten()
for i, col in enumerate(cols):
    sns.boxplot(data=df_data, y=col, ax=axes[i])

plt.show()
```

There is one potential outlier in annual income. However, it seems reasonable to keep this observation.

```python
cols = ["Age", "Annual Income", "Normalized spending"]
sns.pairplot(data=df_data[cols])
plt.show()
```
